```python
# Import libraries
import cv2
import matplotlib.pyplot as plt
```

```python
# Read an image
img_file = r"C:\Users\HP\Downloads\img1.png"
img = cv2.imread(img_file)
```

```python
# Display the image
img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

```python
# Image resize

img_resized_half = cv2.resize(img_rgb, (0,0),fx=0.5,fy=0.5)
img_resized_double = cv2.resize(img_rgb, (0,0),fx=2.0,fy=2.0)

# Image Rotation
img_rotated_90 = cv2.rotate(img_rgb,cv2.ROTATE_90_CLOCKWISE)
img_rotated_180 = cv2.rotate(img_rgb,cv2.ROTATE_180)
img_rotated_270 = cv2.rotate(img_rgb,cv2.ROTATE_90_COUNTERCLOCKWISE)

# Image Flip
img_hflip = cv2.flip(img_rgb,1)
img_vflip = cv2.flip(img_rgb,0)
img_bothflip = cv2.flip(img_rgb,-1)

# Image Cropping
img_crop = img_rgb[0:300,0:300]
```

```python
# Display the images
fig,axes = plt.subplots(3,3,figsize=(10,7))
axes = axes.ravel()
axes[0].imshow(img_rgb)
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(img_resized_half)
axes[1].set_title('Resized Image:half')
axes[1].axis('off')

axes[2].imshow(img_resized_double)
axes[2].set_title('Resized Image:double')
axes[2].axis('off')

axes[3].imshow(img_rotated_90)
axes[3].set_title('Rotated Image 90d')
axes[3].axis('off')

axes[4].imshow(img_rotated_180)
axes[4].set_title('Rotated Image 180d')
axes[4].axis('off')

axes[5].imshow(img_rotated_270)
axes[5].set_title('Rotated Image 270d')
axes[5].axis('off')

axes[6].imshow(img_hflip)
axes[6].set_title('Horizontal Flip')
axes[6].axis('off')

axes[7].imshow(img_vflip)
axes[7].set_title('Vertical Flip')
axes[7].axis('off')

axes[8].imshow(img_crop)
axes[8].set_title('Cropped Image')
axes[8].axis('off')

plt.tight_layout()
plt.show()

```

```python
img_dir = r"D:\jpg format image\img43.png"
img_low = cv2.imread(img_dir)
```

```python
plt.imshow(img_low,cmap='gray')
plt.axis('off')
plt.show()
```

```python
import numpy as np
# Determine the minimum and maximum pixel values in the original image
I_min = np.min(img_low)
I_max = np.max(img_low)
print(I_min,I_max)
```

```python
# Define the new minimum and maximum values
I_min_new = 0
I_max_new = 255
```

```python
img_gray = cv2.cvtColor(img_low,cv2.COLOR_BGR2GRAY)
```

```python
# stretching 
new_img = ((img_low - I_min) * ((I_max_new - I_min_new) / (I_max - I_min)) + I_min_new).astype(np.uint8)
plt.imshow(new_img,cmap='gray')
plt.axis('off')
plt.show()
```
