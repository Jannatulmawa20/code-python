```python
#Image Smoothing
import cv2
import numpy as np
from scipy.ndimage import median_filter
import matplotlib.pyplot as plt
```

```python
# Load an image in grayscale
image = cv2.imread(r"E:\download.png", cv2.IMREAD_GRAYSCALE)
img_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

```python
# Apply Mean Filter (using a 3x3 kernel)
mean_filtered = cv2.blur(img_rgb, (3, 3))

# Apply Median Filter (using a 3x3 kernel)
median_filtered = median_filter(img_rgb, size=3)
```

```python
# Display the original and filtered images
plt.figure(figsize=(10, 5))

# Original Image
plt.subplot(1, 3, 1)
plt.title("Original Image")
plt.imshow(img_rgb, cmap='gray')
plt.axis('off')

# Mean Filtered Image
plt.subplot(1, 3, 2)
plt.title("Mean Filter")
plt.imshow(mean_filtered, cmap='gray')
plt.axis('off')

# Median Filtered Image
plt.subplot(1, 3, 3)
plt.title("Median Filter")
plt.imshow(median_filtered, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
from scipy.ndimage import median_filter
import matplotlib.pyplot as plt
```

```python
# Load an image in grayscale
image = cv2.imread(r"E:\download (1).png", cv2.IMREAD_GRAYSCALE)
img_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

```python
# Apply Sobel Filter
# Sobel X for horizontal edges
sobel_x = cv2.Sobel(img_rgb, cv2.CV_64F, 1, 0, ksize=3)  # ksize=3 is the size of the Sobel kernel; 
# cv2.CV_64F: Specifies a 64-bit float output, allowing for negative gradients. This is essential for accurately capturing the gradient changes.
sobel_x = cv2.convertScaleAbs(sobel_x)  # Convert back to uint8

# Sobel Y for vertical edges
sobel_y = cv2.Sobel(img_rgb, cv2.CV_64F, 0, 1, ksize=3)
sobel_y = cv2.convertScaleAbs(sobel_y)

# Combine Sobel X and Y
sobel_combined = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)
```

```python
# Apply Laplacian Filter
laplacian = cv2.Laplacian(img_rgb, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)
```

```python
# Display the original and filtered images
plt.figure(figsize=(12, 6))

# Original Image
plt.subplot(1, 4, 1)
plt.title("Original Image")
plt.imshow(img_rgb, cmap='gray')
plt.axis('off')

# Sobel X
plt.subplot(1, 4, 2)
plt.title("Sobel X")
plt.imshow(sobel_x, cmap='gray')
plt.axis('off')

# Sobel Y
plt.subplot(1, 4, 3)
plt.title("Sobel Y")
plt.imshow(sobel_y, cmap='gray')
plt.axis('off')

# Sobel Combined
plt.subplot(1, 4, 4)
plt.title("Sobel Combined")
plt.imshow(sobel_combined, cmap='gray')
plt.axis('off')

# Laplacian Image
plt.figure(figsize=(6, 6))
plt.title("Laplacian Filter")
plt.imshow(laplacian, cmap='gray')
plt.axis('off')

plt.show()
```
