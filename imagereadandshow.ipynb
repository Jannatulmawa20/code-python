```python
#read an image from computer path
import cv2
img_path = r"C:\Users\HP\Downloads\img1.png"
img = cv2.imread(img_path)
```

```python
# Display the image characeristics 
print("Dimension(h,w,c) = ",img.shape) 
print("Size = ",img.size)
print("Data Types = ",img.dtype)
```

```python
# Convert from BGR (used by OpenCV) to RGB (used by matplotlib)
img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

```python
#display the image
import matplotlib.pyplot as plt
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

```python
#Generate image list
import os
folder_path = r"D:\jpg format image"
image_files=[] # Create a list that stores all image names as a list

for file_name in os.listdir(folder_path):
    if file_name.endswith('.png'):
        #generate image path
        img_path = os.path.join(folder_path,file_name)
        
        #add it into the list
        image_files.append(img_path)
```

```python
image_files
```

```python
# Create a 2 x 3 subplot 
fig, axes = plt.subplots(2, 3, figsize=(10, 7))
axes = axes.ravel() # flatten the the 2D matrix into 1D so that we can manage the code


img_count = 6
image_files = image_files[:img_count]

for i,img_file in enumerate(image_files):
    img = cv2.imread(img_file)
    img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
    axes[i].imshow(img_rgb)
    axes[i].axis('off')

# Adjust layout and show the plot
plt.tight_layout()
plt.show()
```
